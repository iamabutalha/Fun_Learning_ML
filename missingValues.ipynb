# Imputing Missing Values in python

- Missing values are those values which are not present in dataset like NaN, NA, None
- It can be handle by:
    - `removing`
    - `imputing`
    - `using`

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 


data = sns.load_dataset("titanic")
data.head()
```

```python
# check number of missing values in each column
data.isnull().sum().sort_values(ascending=False) 
```

```python
# drop deck column
data = data.drop('deck', axis=1)
data.head()
```

```python
data.isnull().sum().sort_values(ascending=False) 
```

```python
data['age'] = data['age'].fillna(data['age'].mean())
data.head()
```

```python
data.isnull().sum().sort_values(ascending=False)
```

# Median Imputation

```python
df = sns.load_dataset("titanic")

df['age'] = df['age'].fillna(df['age'].median())

#show first 10 rows
df.head(10)

```

# Mode Categorical Imputation

```python
df = sns.load_dataset("titanic")
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['deck'] = df['deck'].fillna(df['deck'].mode()[0])
df['age'] = df['age'].fillna(round(df['age'].mean(), 0))
```

```python
df.isnull().sum(). sort_values(ascending=False)
df.head(10)
```

# K Nearest Neighbour Method

```python
df = sns.load_dataset('titanic')
df.isnull().sum().sort_values(ascending=False)
```

```python
from sklearn.impute import KNNImputer

# call the KNN with 4 neighbors
imputer = KNNImputer(n_neighbors=4)

# impute the missing values in the 'age' column

#  fit transform means it will first learn from the data and then transform it
# first fit the model on the data then transform the data
# df['age'] = inputer.fit(df[['age']].transform(df[['age']]) can be also done like this
df['age'] = imputer.fit_transform(df[['age']])

df.isnull().sum().sort_values(ascending=False)
```

```python
from sklearn.preprocessing import LabelEncoder

# Column to encode
columns_to_encode = ["sex", "embarked", 'who', 'deck', 'class', 'embark_town', 'alive', 'adult_male']

label_encoders = {}

for col in columns_to_encode:
     le = LabelEncoder()

     df[col] = le.fit_transform(df[col])
     label_encoders[col] = le

df.head()
label_encoders
```

### this below part is KNN imputer which is also regression imputer

```python
# impute the missing values with ItrativeImputer
# it will first run and the iteration mean it will run that much time and will improve what it done wrong

# it will iterate 10 time on one column to correct whole data if it miss in a iteration and then it will find average of that missing value
# from sklearn.impute import IterativeImputer
imputer = IterativeImputer(max_iter=10)
columns_to_inpute = ['age', 'embark_town', 'embarked', 'deck']

for col in columns_to_inpute:
    # the 2D array mean all the columns are contributing to predict the value of the column which is missing
    # it predict value on the basics of all the columns 
    df[col] = imputer.fit_transform(df[[col]])

df.isnull().sum().sort_values(ascending=False)

df.head()
```

```python
# reverse transform the encoded columns to original values
#IM fro reverse encoding i must have to run only labelencoder or imputer on data bcz the other algorithm will make it hard to reverse categorical data bcz it will change the data to float or int

for col in columns_to_encode:
    le = label_encoders[col]
    df[col] = le.inverse_transform(df[col].astype(int))

df.head()
```

# Regression Imputation



```python
# load dataset
df = sns.load_dataset('titanic')

# count null or missing values in each column  
df.isnull().sum().sort_values(ascending=False)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error
from sklearn.impute import SimpleImputer 

df = sns.load_dataset('titanic')

df.isnull().sum().sort_values(ascending=False)
```

```python
# drop deck column
df.drop('deck',axis=1, inplace=True)

df.isnull().sum().sort_values(ascending=False)
```

```python
# label encode data again for RandomForest
from sklearn.preprocessing import LabelEncoder

columns_to_encode = ['sex', 'embarked', 'who', 'class', 'embark_town', 'alive']

# dictionary to stroe label encode of each column
label_encoders = {}

for col in columns_to_encode:
    le = LabelEncoder()
# encode all the above columns
    df[col] = le.fit_transform(df[col])

# store label encoder in dictionary
    label_encoders[col] = le

df.head()
```

```python
# split the dataset into two parts one with missing values and one with no missing values
# getting all the rows where age is missing
df_with_missing = df[df['age'].isna()]

# remove all the rows which have age missing bcz we have put all them into df_with_missing
df_without_missing = df.dropna()

# df_with_missing.head()
# df_without_missing.head()

```

```python
print('The shape of the original dataset is: ', df.shape)
print('The shape of the dataset with missing value is: ', df_with_missing.shape)
print('The shape of the dataset without missing value is: ', df_without_missing.shape)
```

```python
# Regression Imputation  
# Now here we will take all the colums as x i.e features/inputs to predict the missing age colums
from sklearn.metrics import r2_score, mean_absolute_error, mean_absolute_percentage_error, mean_squared_error
# First we will drop age from the without missing values bcz there age is avaliable
# these both values are known to us X and y
# take X as all the rows from dataframe without age 
X = df_without_missing.drop(['age'], axis=1)
# take y as the age column bcz we know that
y = df_without_missing['age']

# Split the data into test train 80% train 20% test
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)

# n_estimator mean how many trees we want to run in random forest regrerssor

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model = rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)
print("RMSE for Random Forest Imputation: ", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score for Random Forest Imputation: ",r2_score(y_test, y_pred))
print("MAE for Random Forest Imputation: ", mean_absolute_error(y_test, y_pred))
print("MAPE for Random Forest Imputation: ", mean_absolute_percentage_error(y_test, y_pred))


```

```python
df_with_missing.isnull().sum().sort_values(ascending=False)
```

```python
# predict the missing values

# here the paramater act as X
# it mean just drop the age which is y from the missing values and predict the age bcz the df_with_missing all other columns are features 
y_pred = rf_model.predict(df_with_missing.drop(['age'], axis=1))
# y_pred
```

```python
# remove warning
import warnings
warnings.filterwarnings('ignore')

# replace the missing values with the predicted values
df_with_missing['age'] = y_pred

df_with_missing.isnull().sum().sort_values(ascending=False)
```

```python
# concate the dataFrames
# now in df_with_missing we have age predicted the age is not missing
df_complete = pd.concat([df_with_missing, df_without_missing], axis=0)
print('The Shape of the complete dataframe is ', df_complete.shape)

df_with_missing.head()
```

```python
# now inverse transform the whole data again to make categorical variable again categorical
for col in columns_to_encode:
    le = label_encoders[col]

    df_complete[col] = le.inverse_transform(df[col])


# df_complete['age'] = round(df_complete['age'])

df_complete.head()

df_complete.to_csv('titanic.csv', index=False)
```
