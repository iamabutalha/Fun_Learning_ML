```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

df = sns.load_dataset('tips')

df.head()
```

```python
df['smoker_num'] = df['smoker'].map({'Yes': 1, 'No': 0})
df.head()
```

```python
X = df[['total_bill', 'size', 'smoker_num']]
Y = df['tip']


model = LinearRegression()

# train/fit model
model.fit(X,Y)
```

```python
print(model.predict([[35.527, 2, 1]]))
print(model.predict([[35.527, 2, 0]]))
```

# Decision Tree Regression



```python
#import decision tree regression
from sklearn.tree import DecisionTreeRegressor
df = sns.load_dataset('tips')

df.head()
X = df[['total_bill', 'size']]
Y = df['tip']

# plot data with best fit line simple one line 
sns.lmplot(x='total_bill', y='tip', data=df, hue='smoker')

model = DecisionTreeRegressor()

# train/fit model
model.fit(X,Y)
print(model.predict([[35.527, 2]]))
print(model.predict([[35.527, 5]]))

```

# Logistic Regression
logistic regression is type of classification not regression
 It is used to predict the probability of a binary outcome (0 or 1)
 It is also called Binary Classification


```python
from sklearn.linear_model import LogisticRegression

df = sns.load_dataset('tips')

df.head()
X = df[['total_bill', 'size', 'tip']]
Y = df['smoker']


model = LogisticRegression()

# train/fit model
model.fit(X,Y)
print(model.predict([[35.527, 2, 3.5]]))
print(model.predict([[35.527, 2, 2]]))

```

# Navie Bayes

```python
from sklearn.naive_bayes import GaussianNB as NaiveBayes
df = sns.load_dataset('tips')

df.head()
X = df[['total_bill', 'size', 'tip']]
Y = df['smoker']


model = NaiveBayes()

# train/fit model
model.fit(X,Y)
print(model.predict([[35.527, 2, 3.5]]))
print(model.predict([[35.527, 2, 2]]))

```
