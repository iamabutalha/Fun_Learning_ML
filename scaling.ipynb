Scaling


### 1.1 Standard scaler

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler
```

```python
df = {
    'age': [25,30, 35, 40, 45],
    'height': [165,170, 175, 180, 185],
    'weight': [55,60, 65,70,75]
}

df = pd.DataFrame(df)
df
```

```python
# import the scaler

scaler = StandardScaler()

#fit scaler on data
scaled_df = scaler.fit_transform(df)
scaled_df

scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df
```

### 1.2 Min Max Scaler

```python
scaler = MinMaxScaler()


scaled_df = scaler.fit_transform(df)
# convert into pandas df

scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df.head()
```

### 1.3 MaxAbsScaler


```python
scaler = MaxAbsScaler()

scaled_df = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df.head()
```

### 1.4 Roboust Scaler

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

scaled_df = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_df, columns=df.columns)

scaled_df.head()
```

## Transformation

we can also do transformation instead of scaling

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# generate the non linear data
np.random.seed(0)
df = np.random.exponential(size=1000, scale=2)
df = pd.DataFrame(df, columns=['values'])
df.head()
```

```python
sns.histplot(df['values'], kde=True)
```

```python
from sklearn.preprocessing import PowerTransformer
from sklearn.preprocessing import QuantileTransformer

pt_boxcox = PowerTransformer(method='box-cox', standardize=False)
pt_yeo_jhonson = PowerTransformer(method='yeo-johnson', standardize=False)
qt_normal = QuantileTransformer(output_distribution='normal')

# boxcox k liye data must be postivie

df['Box-cox'] = pt_boxcox.fit_transform(df[['values']] + 1)
df['Yeo_Jhonson'] = pt_yeo_jhonson.fit_transform(df[['values']])
df['Quantile'] = qt_normal.fit_transform(df[['values']])
df.head()
```

```python
for col in df.columns:
    sns.histplot(df[col], kde=True)
    plt.show()
```
